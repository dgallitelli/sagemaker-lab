# Step 0: Upload Raw Electricity Data to S3

This notebook copies the UCI Electricity dataset (`LD2011_2014.txt.zip`) from the SageMaker example files bucket to your own S3 bucket. No local download is needed — it's a server-side S3 copy.

## Configuration

```python
import boto3
import sagemaker

# --- Edit these as needed ---
REGION = boto3.session.Session().region_name
sess = sagemaker.session.Session()
BUCKET = sess.default_bucket()
S3_PREFIX = "autogluon-timeseries"
S3_OUTPUT = f"s3://{BUCKET}/{S3_PREFIX}/raw/"

print(f"Region:    {REGION}")
print(f"S3 output: {S3_OUTPUT}")
```

## Copy Raw Data from SageMaker Example Bucket

The raw zip file lives in the region-specific SageMaker example bucket. We do a server-side `copy_object` to your bucket.

```python
source_bucket = f"sagemaker-example-files-prod-{REGION}"
filename = "LD2011_2014.txt.zip"
source_key = f"datasets/timeseries/uci_electricity/{filename}"

dest_bucket = BUCKET
dest_key = f"{S3_PREFIX}/raw/{filename}"

s3 = boto3.client("s3", region_name=REGION)

print(f"Copying s3://{source_bucket}/{source_key} -> s3://{dest_bucket}/{dest_key}")
s3.copy_object(
    CopySource={"Bucket": source_bucket, "Key": source_key},
    Bucket=dest_bucket,
    Key=dest_key,
)
print(f"\nRaw data uploaded to s3://{dest_bucket}/{S3_PREFIX}/raw/")
print("Next: run 1_launch_processing.ipynb to preprocess with SageMaker Processing.")
```
