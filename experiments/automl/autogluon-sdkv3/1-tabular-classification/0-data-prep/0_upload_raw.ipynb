# Step 0: Upload Raw Data to S3

Download raw UCI Adult Census data from SageMaker example files and upload to your S3 bucket.
This is the only local step — everything else runs on SageMaker.

## Configuration

```python
import boto3
import sagemaker

REGION = boto3.session.Session().region_name
sess = sagemaker.session.Session()
BUCKET = sess.default_bucket()
S3_PREFIX = "autogluon-tabular"
S3_OUTPUT = f"s3://{BUCKET}/{S3_PREFIX}/raw/"
```

## Copy raw data from SageMaker example files to S3

```python
source_bucket = f"sagemaker-example-files-prod-{REGION}"
s3 = boto3.client("s3", region_name=REGION)

# Parse destination
dest = S3_OUTPUT.replace("s3://", "")
dest_bucket = dest.split("/")[0]
dest_prefix = "/".join(dest.split("/")[1:]).rstrip("/")

for filename in ("adult.data", "adult.test"):
    source_key = f"datasets/tabular/uci_adult/{filename}"
    dest_key = f"{dest_prefix}/{filename}"
    print(f"Copying s3://{source_bucket}/{source_key} -> s3://{dest_bucket}/{dest_key}")
    s3.copy_object(
        CopySource={"Bucket": source_bucket, "Key": source_key},
        Bucket=dest_bucket,
        Key=dest_key,
    )

print(f"\nRaw data uploaded to s3://{dest_bucket}/{dest_prefix}/")
print("Next: run 1_launch_processing.ipynb to preprocess with SageMaker Processing.")
```
